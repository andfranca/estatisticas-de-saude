## **Tabulação dos Dados**
Ao tentar acessar o tabnet a página estava indisponível, então optei por seguir de outra maneira utilizando a base de dados completa do SINASC

```python
#Tabulaçaõ dos Dados
import pandas as pd
import numpy as np

df = pd.read_csv('/content/SINASC_2024.csv',
                 low_memory=False,
                 sep=';')
```

```python
# Filtrar Ribeirão Preto e Registro
municipios = {354340: 'Ribeirão Preto',
              354260: 'Registro'}
df_analise = df[df['CODMUNRES'].astype(int).isin(municipios.keys())].copy()
```

```python
# Tratamento da variável PESO (gramas)
# Converter para numérico e tratar nulos (RIPSA recomenda excluir 'ignorado' do denominador)
df_analise['PESO'] = pd.to_numeric(df_analise['PESO'], errors='coerce')
df_analise = df_analise.dropna(subset=['PESO'])
```

```python
# Definir Baixo Peso ao Nascer (BPN < 2500g)
df_analise['BPN'] = df_analise['PESO'] < 2500
```

```python
# Calcular da Proporção por Município
resultado = df_analise.groupby('CODMUNRES')['BPN'].agg(['sum', 'count']).reset_index()
resultado['prop_bpn'] = (resultado['sum'] / resultado['count']) * 100
resultado['municipio'] = resultado['CODMUNRES'].map(municipios)
```

## **Resultado**

```python
print(resultado[['municipio', 'sum', 'count', 'prop_bpn']])
```

## **Análise**

### **Resultados Obtidos**

A partir da tabulação dos dados do SINASC (2024), calculamos a proporção de bebês nascidos com menos de $2.500g$ em relação ao total de nascidos vivos com peso conhecido:

- Ribeirão Preto: 10,09%
- Registro: 7,37%


### **Interpretação dos Níveis**

De acordo com a RIPSA, o baixo peso ao nascer acontece basicamente por dois motivos:

- **Prematuridade:** Quando o bebê nasce antes da hora (antes das 37 semanas).
- **Retardo de crescimento intrauterino:** Quando o bebê nasce no tempo certo, mas "magrinho" ou pequeno demais, porque não se desenvolveu como deveria durante a gestação.

A ficha técnica da RIPSA afirma que valores acima de 10% são considerados inaceitáveis segundo padrões internacionais. Ribeirão Preto (10,08%) atinge esse patamar crítico. Já Registro (7,36%) apresenta um valor mais próximo do observado em países desenvolvidos (5-6%), embora ainda acima deles.

A ficha (item 4) alerta que, em áreas menos desenvolvidas, valores baixos podem representar sub-registro ou baixa qualidade da informação. No caso de Registro, seria necessário verificar se todos os nascimentos estão sendo captados corretamente pelo SINASC para garantir que os 7,36% são reais e não fruto de dados faltantes.
